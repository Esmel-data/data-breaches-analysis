## 📚 1. Import des librairies

```python
import requests
import pandas as pd
import os
import time
from datetime import datetime

print("✅ Librairies importées avec succès")
print(f"📅 Date d'exécution : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

## 📁 2. Création des dossiers du projet

## 🌐 3. Collecte des données via l'API HIBP

> **API utilisée** : `https://haveibeenpwned.com/api/v3/breaches`  
> **Authentification** : Aucune clé requise pour la liste des breaches  
> **Format** : JSON

```python
# Configuration de l'API
API_URL = "https://haveibeenpwned.com/api/v3/breaches"
HEADERS = {
    "User-Agent": "DataAnalysis-Portfolio-EsmelAmari"
}

print("🔄 Connexion à l'API HaveIBeenPwned...")

try:
    response = requests.get(API_URL, headers=HEADERS, timeout=15)
    
    # Vérification du statut
    if response.status_code == 200:
        data = response.json()
        print(f"✅ Connexion réussie !")
        print(f"📊 Nombre total de fuites récupérées : {len(data)}")
    else:
        print(f"❌ Erreur HTTP : {response.status_code}")
        print(response.text)

except requests.exceptions.ConnectionError:
    print("❌ Erreur de connexion. Vérifiez votre accès internet.")
except requests.exceptions.Timeout:
    print("❌ Délai dépassé. L'API ne répond pas.")
except Exception as e:
    print(f"❌ Erreur inattendue : {e}")
```

## 🔍 4. Inspection brute des données

```python
# Aperçu du premier enregistrement brut (format JSON)
print("🔎 Exemple d'un enregistrement brut (1er breach) :")
print("-" * 60)
for key, value in data[0].items():
    print(f"  {key:20s} : {value}")
```

```python
# Conversion en DataFrame
df = pd.DataFrame(data)

print(f"📐 Dimensions du DataFrame : {df.shape[0]} lignes x {df.shape[1]} colonnes")
print(f"\n📋 Liste des colonnes :")
for i, col in enumerate(df.columns, 1):
    print(f"  {i:02d}. {col}")
```

```python
# Aperçu des 5 premières lignes
print("📄 Aperçu des 5 premières lignes :")
df.head()
```

```python
# Types de données
print("🧬 Types de données par colonne :")
print("-" * 40)
print(df.dtypes)
```

```python
# Valeurs manquantes
print("🕳️  Valeurs manquantes par colonne :")
print("-" * 40)
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Manquants': missing, 'Pourcentage (%)': missing_pct})
print(missing_df[missing_df['Manquants'] > 0])

if missing.sum() == 0:
    print("✅ Aucune valeur manquante détectée")
```

## 📊 5. Statistiques rapides sur les données brutes

```python
# Statistiques sur PwnCount (nombre de comptes compromis)
print("💥 Statistiques sur le nombre de comptes compromis (PwnCount) :")
print("-" * 55)
print(f"  Total comptes compromis  : {df['PwnCount'].sum():,.0f}")
print(f"  Moyenne par fuite        : {df['PwnCount'].mean():,.0f}")
print(f"  Maximum (1 seule fuite)  : {df['PwnCount'].max():,.0f}")
print(f"  Minimum                  : {df['PwnCount'].min():,.0f}")
print(f"  Médiane                  : {df['PwnCount'].median():,.0f}")
```

```python
# Top 5 des plus grandes fuites
print("🏆 Top 5 des plus grandes fuites (par nombre de comptes) :")
print("-" * 60)
top5 = df[['Name', 'PwnCount', 'BreachDate', 'Domain']].sort_values('PwnCount', ascending=False).head(5)
top5['PwnCount_M'] = (top5['PwnCount'] / 1_000_000).round(1).astype(str) + 'M'
print(top5[['Name', 'PwnCount_M', 'BreachDate', 'Domain']].to_string(index=False))
```

```python
# Exploration de la colonne DataClasses
print("📂 Aperçu des types de données exposées (DataClasses) :")
print("-" * 50)

# Aplatir toutes les DataClasses
all_dataclasses = [item for sublist in df['DataClasses'].dropna() for item in sublist]
dataclasses_series = pd.Series(all_dataclasses)

print(f"  Nombre total de types uniques : {dataclasses_series.nunique()}")
print(f"\n  Top 10 types de données les plus exposés :")
print(dataclasses_series.value_counts().head(10).to_string())
```

```python
# Colonnes booléennes - répartition
bool_cols = ['IsVerified', 'IsFabricated', 'IsSensitive', 'IsRetired', 'IsSpamList', 'IsMalware']
bool_cols_present = [c for c in bool_cols if c in df.columns]

print("🔘 Répartition des colonnes booléennes :")
print("-" * 45)
for col in bool_cols_present:
    true_count = df[col].sum()
    pct = round(true_count / len(df) * 100, 1)
    print(f"  {col:20s} : {true_count:4d} ({pct}%)")
```

## 💾 6. Export des données brutes en CSV

```python
# Export CSV brut
output_path = '../data/breaches_raw.csv'
df.to_csv(output_path, index=False, encoding='utf-8')

# Vérification
file_size = os.path.getsize(output_path) / 1024
print("💾 Export réussi !")
print("-" * 45)
print(f"  📄 Fichier     : {output_path}")
print(f"  📐 Dimensions  : {df.shape[0]} lignes x {df.shape[1]} colonnes")
print(f"  💿 Taille      : {file_size:.1f} KB")
print(f"  📅 Exporté le  : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```
